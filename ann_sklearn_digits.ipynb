# Digit Classification with SVM

##Install libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

## 1. Data Loading

```python
digit=load_digits()
```

## 2. Data Preparation

```python
x,y=digit.data,digit.target
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.5,random_state=42)
```

## 3. Model Training

```python
model=SVC(gamma=0.001)
model.fit(X_train,y_train)

pred=model.predict(X_test)
```

## 4. Model Evaluation

```python
accuracy=accuracy_score(y_test,pred)
```

```python
print(f' Accuracy is : {accuracy * 100:.2f}%')
```

## 5. Visualization of Predictions

```python
import matplotlib.pyplot as plt

_,axes=plt.subplots(1,4, figsize=(10,3))

for ax,img,prediction,real in zip(axes,X_test,pred,y_test):
    ax.axis('off')
    s=img.reshape(8,8)
    ax.imshow(s,cmap='gray')
    c='green' if prediction==real else 'red'
    ax.set_title(f'pred:{prediction}\nreal:{real}',color=c)

plt.show()


```
